```python
import pandas as pd

df = pd.read_csv("Global YouTube Statistics.csv", encoding='latin1')

print(df.head())
```

```python
print('Summary Statistics for Numerical Columns:')
display(df.describe())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x="subscribers",
    y="video views",
    color="blue",
    alpha=0.6
)

plt.title("Subscribers vs. Video Views")
plt.xlabel("Subscribers")
plt.ylabel("Video Views")
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.boxplot(x=df["subscribers"], color="orange")

plt.title("Subscribers Outlier Detection")
plt.xlabel("Subscribers")

plt.show()
```

```python
display(df.head())
```

```python
correlation = df['subscribers'].corr(df['video views'])
print(f"Correlation between Subscribers and Video Views: {correlation:.4f}")
```

```python
display(df.describe())
```

```python
mean_subscribers_by_category = df.groupby('category')['subscribers'].mean().sort_values(ascending=False)
display(mean_subscribers_by_category)
```

```python
total_video_views_by_category = df.groupby('category')['video views'].sum().sort_values(ascending=False)
display(total_video_views_by_category)
```

```python
subscribers_by_year = df.groupby('created_year')['subscribers'].mean().reset_index()

plt.figure(figsize=(12,5))

sns.lineplot(
    data=subscribers_by_year,
    x='created_year',
    y='subscribers',
    color='green'
)

plt.title('Average Subscribers Over Created Year')

plt.xlabel('Created Year')
plt.ylabel('Average Subscribers')

plt.grid(True, linestyle='--', alpha=0.7)
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
