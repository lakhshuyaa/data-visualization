```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Global YouTube Statistics.csv", encoding='latin1')
```

```python
print(df.head())

print(df.shape)

print(df.dtypes)
```

```python
print(df.describe())
```

```python
high_aqi = df[df["AQI"] > 300]

print(high_aqi.head())
```

```python
print(df.columns)
```

```python
display(df.head())
```

```python
delhi = df[df["Country"] == "India"]

print(delhi.head())
```

```python
sorted_df = df.sort_values(
    by="Country",
    ascending=False
)

print(sorted_df.head())
```

```python
print("Mean :", df["subscribers"].mean())

print("Median :", df["subscribers"].median())

print("Mode :")

print(df["subscribers"].mode())

print("Standard Deviation :", df["subscribers"].std())
```

```python
country_avg = df.groupby("Country")["subscribers"].mean()

print(country_avg.sort_values(ascending=False).head(10))
```

```python
country_avg.sort_values(ascending=False).head(10).plot(kind="bar", figsize=(12, 6))

plt.title("Top 10 Countries by Average Subscribers")

plt.ylabel("Average Subscribers")

plt.xlabel("Country")

plt.xticks(rotation=45, ha='right')

plt.tight_layout()

plt.show()
```

```python
import numpy as np

# Calculate the correlation matrix for numerical columns
corr = df.select_dtypes(include=np.number).corr()
```

```python
plt.figure(figsize=(15,10))
sns.heatmap(corr,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")
plt.title("Correlation Heatmap of Numerical Columns")
plt.show()
```
